````python
f = '/data/docker/lamnguyenx/kinh/giao_ly_questions.txt'
Lines = [l.strip() for l in open(f, 'r').readlines() if l.strip()]
Lines[:10]
````

````python
f_answers = '/data/docker/lamnguyenx/kinh/giao_ly_questions-answers.txt'

Aines = [l.strip() for l in open(f_answers, 'r').readlines() if len(l.strip())>0]
print(Aines[-10:])
print(int(len(Aines)/2))

Q2A = {}
for i in range(0, int(len(Aines)), 2):
    # print(Aines[i], Aines[i+1])
    Q2A[int(Aines[i])] = Aines[i+1]

Q2A = dict(sorted(list(Q2A.items())))
len(Q2A), Q2A
````

````python
Questions = []
counter = 0
for line in Lines:
    if line.startswith('Câu '):
        Q_num = int(line.split(':', maxsplit=1)[0].split()[-1])
        Questions.append({'Question' : line.split(':', maxsplit=1)[-1].strip()})
        Questions[-1]['num'] = Q_num
        Questions[-1]['ans'] = Q2A[Q_num]

    if line.startswith(r'£'):
        line = line.removeprefix(r'£ ')
        q_index = line[0]
        line = line[3:]
        Questions[-1][q_index] = line

Questions
````

````python
import pandas
DF = pandas.DataFrame(Questions)
DF
````

````python
f_out = '/data/docker/lamnguyenx/kinh/TRAC_NGHIEM.01.MD'
with open(f_out, 'w+') as F:
    for i, Row in DF.iterrows():

        line = f'#### Câu {Row.loc['num']}: {Row.loc['Question']}'
        F.write(line + '\n'*2)
        for i, l in enumerate(('A', 'B', 'C', 'D')):
            if l == Row.loc['ans']:
                F.write(f'_{l}. {Row.loc[l]}_\n\n')
            else:
                F.write(f'{l}. {Row.loc[l]}\n\n')
````
