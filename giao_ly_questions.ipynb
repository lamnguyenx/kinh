```python
f = '/data/docker/lamnguyenx/kinh/giao_ly_questions.txt'
Lines = [l.strip() for l in open(f, 'r').readlines() if l.strip()]
Lines[:10]
```

```python
Questions = []
counter = 0
for line in Lines:
    if line.startswith('Câu '):
        Questions.append({'Question' : line.split(':', maxsplit=1)[-1].strip()})

    if line.startswith(r'£'):
        line = line.removeprefix(r'£ ')
        q_index = line[0]
        line = line[3:]
        Questions[-1][q_index] = line

Questions[:2]
```

```python
import pandas
DF = pandas.DataFrame(Questions)
DF
```
